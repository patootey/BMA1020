## 3. Ray tracing

Remember the form factors we calculated for use in the radiosity algorithm? Our idea is (as indicated in the lectures) to compute these by a ray-tracing strategy, i.e. with simulation. This will automatically take into account visibility.

*Various different, but related methods take the name ray tracing. We are staying in the radiosity/diffuse setting, but it is not so difficult to modify the techniques below to account for specular reflection etc.*

The idea is as follows. For each element $i$:
1. Generate $n$ rays from random points $x$ inside $i$, in random directions $n$ (following a Lambertian distribution)
2. Find out which rays hit which element.
3. Add $\frac{1}{n}$ to $F_{ij}$ for each ray that hits element $j$.
